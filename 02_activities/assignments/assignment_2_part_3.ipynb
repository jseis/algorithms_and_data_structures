Take as input a string of bracket characters in the set {'(', ')', '[', ']', '{', '}'}. Determine whether each open bracket is closed by a bracket of the same type and the sequence of open/closed brackets observes a proper nesting structure.


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


My example: "[[(])]"

The function should evaluate False for this input since a single ']' is enclosed within a pair of round brackets, meaning the sequence is not closed in the correct order.

My partner created the following example: "({})"

Working from the beginning of the string, a '(' is encountered first, followed by a '{'. The '{' must be closed prior to encountering either a ')' or a ']', which is the case. Finally, a ')' closes the first character, '('. This should evaluate to True.


-   Copy the solution your partner wrote. 


```python
# Your answer here
def is_valid_brackets(s: str) -> bool:
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []

    for ch in s:
        if ch in pairs.values():          
            stack.append(ch)
        else:                             
            if not stack:
                return False
            top = stack.pop()
            if pairs.get(ch) != top:
                return False

    return len(stack) == 0
```

```python
assert is_valid_brackets("([]{})") == True
assert is_valid_brackets("([)]") == False
assert is_valid_brackets("()[]{}") == True
assert is_valid_brackets("[{]}") == False
```


-   Explain why their solution works in your own words.


The solution iterates from the beginning of the string to the end. When an open bracket is encountered, as identified through the values of a dictionary, it is pushed onto a stack. Otherwise, the character is a closed bracket (assuming the input assumption is not violated). In this case, a corresponding open bracket should already exist in the stack. If the stack is empty, a False is returned. Furthermore, the corresponding open bracket should be the most recently encountered one, or else the sequence would be closed in the wrong order. To check this, the top (most recently added) value is popped off the stack and the dictionary is consulted to determine if it is a match for the closing bracket. If it is not, a False is returned. Once all characters are processed, the function returns a True if there are no open brackets remaining unclosed in the stack and a False otherwise.


-   Explain the problem’s time and space complexity in your own words.


The problem's time and space complexity are each $O(n)$, where $n$ is the length of the input string.

Time complexity: for each character, a single loop iteration consisting of only constant-time operations is run. Checking the fixed dictionary values is $O(1)$, as are pushing and popping a stack.

Space complexity: a stack is used to store previously seen open brackets. This stack could grow as large as $n$, if the input consists of only open brackets.
